```python
import numpy as np
import matplotlib.pyplot as plt

dino = np.load("DINOv3_discovery_curve.npz")
byol = np.load("BYOL_discovery_curve.npz")

dino_x = dino["x"]
dino_recall = dino["recall"]
dino_auc = float(dino["auc"])

byol_x = byol["x"]
byol_recall = byol["recall"]
byol_auc = float(byol["auc"])

print("DINOv3 AUC:", dino_auc)
print("BYOL AUC:", byol_auc)
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    dino_x,
    dino_recall,
    label=f"DINOv3 (AUC = {dino_auc:.3f})"
)

plt.plot(
    byol_x,
    byol_recall,
    label=f"BYOL (AUC = {byol_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random"
)

plt.xlabel("Fraction of ranked sources inspected")
plt.ylabel("Fraction of true anomalies discovered")
plt.title("Discovery Curves")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
plt.savefig("discovery_curves.pdf", bbox_inches="tight")
plt.savefig("discovery_curves.png", dpi=300, bbox_inches="tight")
plt.show()
```
